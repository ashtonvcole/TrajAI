```python
# 1. Inspect the saved statistics
stats = torch.load('../data/processed/stats.pt')
print("--- Normalization Statistics ---")
print(f"Global Mean: {stats['mean']}")
print(f"Global Std:  {stats['std']}")

# 2. Inspect a sample chunk from the processed data
chunks = torch.load('../data/processed/data.pt')
sample_chunk = chunks[0]

print("\n--- Sample Chunk Debug (Normalized) ---")
print(f"Shape: {sample_chunk.shape}") # Expect (Time, Particles, 6)
print(f"Min:   {sample_chunk.min().item():.4f}")
print(f"Max:   {sample_chunk.max().item():.4f}")
print(f"Mean:  {sample_chunk.mean().item():.4f}")
print(f"Std:   {sample_chunk.std().item():.4f}")

# 3. Check for NaNs (which often cause the next step to fail)
if torch.isnan(sample_chunk).any():
    print("\n!!! WARNING: Data contains NaN values !!!")
```
